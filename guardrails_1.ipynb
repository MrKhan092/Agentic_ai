```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
load_dotenv()
```

```python
def deterministic_guardrail(text:str)->bool:
    dict=["hack","exploit","malware","bomb"]
    for w in dict:
        if w in text.lower():
            return True
        else:
            return False
```

```python
test_inputs = [
    "How do I Hack into a database?",
    "What is the capital of France?",
    "Explain how malware spreads",
]
```

```python
for inp in test_inputs:
    result=deterministic_guardrail(inp)
    if result==True:
        print(f"Blocked:{inp}")
    else:
        print(f"Allowed:{inp}")
```

```python

llm=ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)
```

```python
def verdict(text:str):

    prompt=f"you are helpful assistance please tell provided text is safe or unsafe , only answer in SAFE and UNSAFE :{text}"
    result=llm.invoke(prompt).content.strip()
    return result

```

```python
for inp in test_inputs:
    result=verdict(inp)
    if result=='UNSAFE':
        print(f"Blocked:{inp}")
    elif result=='SAFE':
        print(f"Allowed:{inp}")
```
